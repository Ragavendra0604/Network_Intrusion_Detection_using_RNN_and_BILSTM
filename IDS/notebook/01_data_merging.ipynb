```python
# CICIDS 2017 - DATA MERGING & CLEANING

import os
import pandas as pd
import numpy as np
from tqdm import tqdm

# Path to dataset
data_path = r"D:\My_Space\College\6th_SEM\Deep_Learning_Project\IDS\Data\raw\MachineLearningCVE"

# Get all CSV files
files = [f for f in os.listdir(data_path) if f.endswith(".csv")]

print("Found files:")
for f in files:
    print(f)

# Load and clean each file
df_list = []

for file in tqdm(files):
    file_path = os.path.join(data_path, file)
    
    print(f"\nLoading: {file}")
    
    df = pd.read_csv(file_path, low_memory=False)
    
    # Remove duplicate columns spaces
    df.columns = df.columns.str.strip()
    
    # Replace infinity with NaN
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    
    # Drop NaN rows
    df.dropna(inplace=True)
    
    df_list.append(df)

# Merge all files
print("\nMerging all files...")
data = pd.concat(df_list, ignore_index=True)

print("Final shape:", data.shape)

# Remove duplicate rows
data.drop_duplicates(inplace=True)

print("After removing duplicates:", data.shape)

# Save cleaned dataset
save_path = r"D:\My_Space\College\6th_SEM\Deep_Learning_Project\IDS\Data\processed"
os.makedirs(save_path, exist_ok=True)
data.to_csv(os.path.join(save_path, "cicids2017_merged.csv"), index=False)

print("\nDataset saved successfully at:", os.path.join(save_path, "cicids2017_merged.csv"))
```
